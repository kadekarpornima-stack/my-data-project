```python
import os
import glob
import requests
import pandas as pd

# Define path to raw files directory
folder_path = os.path.join('data', 'raw')

# =====================================================================
# TASK 1: LOAD & INSPECT ALL LOCAL CSV FILES
# =====================================================================
print("==================================================")
print("      TASK 1: INSPECTING THE LOCAL DATASETS       ")
print("==================================================\n")

# Grab all available CSV files, ignoring previous live data downloads
csv_files = sorted(glob.glob(os.path.join(folder_path, '*.csv')))
csv_files = [f for f in csv_files if "live_nav" not in os.path.basename(f)]

print(f"🔍 Found {len(csv_files)} baseline datasets to inspect.\n")

for i, file_path in enumerate(csv_files, 1):
    file_name = os.path.basename(file_path)
    print("=" * 60)
    print(f"📄 DATASET {i}: {file_name}")
    print("=" * 60)
    
    try:
        df = pd.read_csv(file_path)
        
        print(f"1. SHAPE (Rows, Columns): {df.shape}")
        print("\n2. DATA TYPES (.dtypes):")
        print(df.dtypes)
        print("-" * 40)
        print("3. DATA PREVIEW (.head()):")
        display(df.head(5))
        
        # Scan for Missing Values / Duplicates
        missing_values = df.isnull().sum().sum()
        duplicate_rows = df.duplicated().sum()
        
        if missing_values > 0 or duplicate_rows > 0:
            print("\n⚠️ POTENTIAL ANOMALIES DETECTED:")
            if missing_values > 0:
                print(f"   - Found {missing_values} missing (NaN) values.")
                if "sip_inflows" in file_name:
                    print("     👉 Note: Missing values are the first 12 rows of 'yoy_growth_pct' due to missing 2021 baseline math.")
            if duplicate_rows > 0:
                print(f"   - Found {duplicate_rows} duplicate rows.")
        else:
            print("\n✅ Structure looks clean! No missing data or duplicates.")
            
    except Exception as e:
        print(f"❌ Error loading {file_name}: {e}")
    print("\n\n")


# =====================================================================
# TASK 2: FETCH LIVE NAV FROM API & SAVE AS RAW CSV
# =====================================================================
print("==================================================")
print("      TASK 2: FETCHING LIVE NAV FROM API         ")
print("==================================================\n")

scheme_id = "125497"  # HDFC Top 100 Direct Plan
api_url = f"https://api.mfapi.in/mf/{scheme_id}"

print(f"📡 Requesting real-time records from: {api_url}")

try:
    # 1. Fetch live JSON payload from API
    response = requests.get(api_url, timeout=15)
    response.raise_for_status()
    
    json_payload = response.json()
    metadata = json_payload.get('meta', {})
    historical_nav_data = json_payload.get('data', [])
    
    fund_name = metadata.get('scheme_name', 'HDFC Mutual Fund')
    print(f"✅ Success! Connected to API for: {fund_name}")
    print(f"📊 Retrieved {len(historical_nav_data)} total timeline entries.")
    
    # 2. Transform into a Pandas DataFrame
    df_live = pd.DataFrame(historical_nav_data)
    df_live['nav'] = pd.to_numeric(df_live['nav'])
    df_live['date'] = pd.to_datetime(df_live['date'], format='%d-%m-%Y')
    df_live = df_live.sort_values('date').reset_index(drop=True)
    
    # 3. Save directly to your data/raw folder
    output_filename = os.path.join(folder_path, f"{scheme_id}_live_nav.csv")
    df_live.to_csv(output_filename, index=False)
    print(f"💾 File safely saved to: {output_filename}")
    
    print(f"\n📊 Live NAV Dataset Final Shape: {df_live.shape}")
    print("\n📈 Top 5 rows of newly fetched data (.head()):")
    display(df_live.head(5))

except Exception as err:
    print(f"❌ API data processing sequence failed: {err}")
```

```python
import os
import time
import requests
import pandas as pd

# 1. Map names to the 5 requested scheme IDs
target_schemes = {
    "sbi_bluechip": "119551",
    "icici_bluechip": "120503",
    "nippon_large_cap": "118632",
    "axis_bluechip": "119092",
    "kotak_bluechip": "120841"
}

output_dir = os.path.join('data', 'raw')
os.makedirs(output_dir, exist_ok=True)

print("🚀 Starting bulk historical NAV acquisition pipeline...\n")

# 2. Iterate sequentially through each fund setup
for fund_name, scheme_id in target_schemes.items():
    api_url = f"https://api.mfapi.in/mf/{scheme_id}"
    print(f"📡 Requesting tracking records for: {fund_name.upper()} (ID: {scheme_id})")
    
    try:
        response = requests.get(api_url, timeout=15)
        response.raise_for_status()
        
        json_payload = response.json()
        historical_data = json_payload.get('data', [])
        
        if not historical_data:
            print(f"⚠️ Warning: No data returned for {fund_name}")
            continue
            
        # 3. Structural conversion into dataframes
        df = pd.DataFrame(historical_data)
        df['nav'] = pd.to_numeric(df['nav'])
        df['date'] = pd.to_datetime(df['date'], format='%d-%m-%Y')
        df = df.sort_values('date').reset_index(drop=True)
        
        # 4. Save to target raw directory
        filename = f"{scheme_id}_{fund_name}_live_nav.csv"
        file_path = os.path.join(output_dir, filename)
        df.to_csv(file_path, index=False)
        
        print(f"✅ Success! Saved {len(df)} days of records to: {file_path}")
        print(f"📈 Quick Preview: Start Date: {df['date'].min().strftime('%Y-%m-%d')} | Latest NAV: {df['nav'].iloc[-1]}")
        print("-" * 70)
        
        # 5. API Etiquette cooldown to protect the free endpoint
        time.sleep(1)
        
    except Exception as err:
        print(f"❌ Failed to acquire data for {fund_name}: {err}")
        print("-" * 70)

print("\n🎉 Bulk ingestion cycle complete! All datasets safely generated.")
```

```python
import os
import pandas as pd

# Define files and their corresponding fund labels
files_to_merge = {
    "119551_sbi_bluechip_live_nav.csv": "SBI_Bluechip",
    "120503_icici_bluechip_live_nav.csv": "ICICI_Bluechip",
    "118632_nippon_large_cap_live_nav.csv": "Nippon_Large_Cap",
    "119092_axis_bluechip_live_nav.csv": "Axis_Bluechip",
    "120841_kotak_bluechip_live_nav.csv": "Kotak_Bluechip"
}

raw_dir = os.path.join('data', 'raw')
master_df = None

print("🔄 Starting data consolidation layer...")

for filename, fund_label in files_to_merge.items():
    file_path = os.path.join(raw_dir, filename)
    
    if os.path.exists(file_path):
        # Read the file and parse dates
        df = pd.read_csv(file_path)
        df['date'] = pd.to_datetime(df['date'])
        
        # Keep only date and nav, renaming nav to the specific fund name
        df = df[['date', 'nav']].rename(columns={'nav': fund_label})
        
        # Merge sequentially
        if master_df is None:
            master_df = df
        else:
            master_df = pd.merge(master_df, df, on='date', how='outer')
            
        print(f"🔹 Integrated {fund_label}")

# Sort from oldest to newest dates
master_df = master_df.sort_values('date').reset_index(drop=True)

# Fill missing values (weekends/holidays) using Forward Fill
master_df = master_df.ffill()

# Save to your 'processed' data folder
processed_dir = os.path.join('data', 'processed')
os.makedirs(processed_dir, exist_ok=True)
output_path = os.path.join(processed_dir, 'consolidated_large_cap_nav.csv')
master_df.to_csv(output_path, index=False)

print(f"\n🎉 Master dataset successfully created at: {output_path}")
print(f"📊 Dimensions: {master_df.shape[0]} rows x {master_df.shape[1]} columns")
master_df.tail(10)
```
